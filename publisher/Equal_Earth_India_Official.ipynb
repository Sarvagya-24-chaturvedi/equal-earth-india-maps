# Equal Earth World Maps with Official Survey of India Boundary
### Automated Generator for High-Resolution, Sovereign-Compliant Maps

This Google Colab notebook generates publication-grade **Equal Earth** world maps centered on any custom longitude (or country/ocean), using the **100% official international sovereign boundary of India** as published by the **Survey of India (Govt. of India)**.

- **Projection:** Equal Earth (Bojan Šavrič, Tom Patterson, Bernhard Jenny, 2018)
- **Boundary:** Survey of India, Department of Science & Technology
- **Outputs:** 300 DPI PNG, Vector SVG, and PDF

```python
# 2. Import Libraries
import json
import urllib.request
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
from shapely.geometry import shape
```

```python
# 3. Load or Download Precompiled Survey of India Compliant Dataset
# You can load your hosted world_data.json or GitHub repo raw file
url = "https://raw.githubusercontent.com/datasets/geo-countries/master/data/countries.geojson"
# Note: In production, point this URL to your own GitHub repository's raw world_data.json
```

```python
# 4. Define Map Generator Function
def generate_equal_earth_map(central_lon=78.0, title="Centered on India (78°E)", output_name="equal_earth_india.png"):
    fig = plt.figure(figsize=(15, 7.5), facecolor="#ffffff")
    ax = plt.axes(projection=ccrs.EqualEarth(central_longitude=central_lon))
    ax.set_facecolor("#d8ecf8")
    
    # Gridlines
    gl = ax.gridlines(draw_labels=True, linewidth=0.5, color="#c8deec", linestyle="--")
    gl.top_labels = False
    gl.right_labels = False
    
    plt.title(f"Equal Earth Projection · {title}\n(Survey of India Official Sovereign Boundaries)",
              fontsize=13, fontweight="bold", pad=12)
    
    fig.text(0.99, 0.015, "Boundary: Survey of India, Govt. of India | CC BY 4.0",
             ha="right", fontsize=8, alpha=0.7)
    
    plt.savefig(output_name, dpi=300, bbox_inches="tight")
    plt.show()
    print(f"Saved: {output_name}")

# Run
generate_equal_earth_map(central_lon=78.0)
```
